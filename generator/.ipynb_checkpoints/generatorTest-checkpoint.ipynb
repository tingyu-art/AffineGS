## 1.Test the flow functions

```python
import torch
from utils.visualizer import plot_flow_field

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from generator.flow import FlowManager

def test_flow():
    # basic test 1
    flow1 = FlowManager(mode="data", data_path="./data/backstep_Re1000_00004.npz")
    flow2 = FlowManager(mode="analytical", name="uniform", uc=1.5,vc=0.05)
    flow3 = FlowManager(mode="analytical", name="solid_rot", x_c=128,y_c=128, omega=0.05)
    flow4 = FlowManager(mode="analytical", name="sin_flow", x_c=128,y_c=128, a=6, b=128, scale=5)
    flow5 = FlowManager(mode="analytical", name="cellular_flow", vmax=10, p=256)
    
    for flow in [flow1, flow2,flow3,flow4,flow5]:
        flow_func = flow.evaluate
    
        # x_vals = torch.linspace(-256, 511, steps=256*3)
        # y_vals = torch.linspace(-256, 511, steps=256*3)
        x_vals = torch.linspace(0, 256, steps=257)
        y_vals = torch.linspace(0, 256, steps=257)
        x, y = torch.meshgrid(x_vals, y_vals, indexing='ij')
        
        u, v = flow_func(x+0.5, y+0.2)
        plot_flow_field(u,v)
    
        # fig = plt.figure()
        # plt.quiver(y[::32,::32].cpu().detach().numpy(),x[::32,::32].cpu().detach().numpy(),
        #            v[::32,::32].cpu().detach().numpy(),-u[::32,::32].cpu().detach().numpy())
        # plt.gca().invert_yaxis()

test_flow()
```

### 2.2.Test the image sequence

```python
from generator.flow import FlowManager
from generator.images import ParticleSequenceSimulator, config_sequence
from utils.visualizer import plot_piv_image_pair, save_piv_video

def test_PSS():
    flow = FlowManager(mode="data", data_path="./data/backstep_Re1000_00004.npz")
    # flow = FlowManager(mode="analytical", name="uniform", uc=1.5,vc=0.00)
    flow_func = flow.evaluate
    
    cfg = config_sequence()
    cfg.d = 2.2
    pss = ParticleSequenceSimulator(flow_func=flow_func,cfg=cfg)
    
    imgs, ts = pss.forward()
    print(f"Generated {len(ts)} timestamps and {len(imgs)} images.")

    plot_piv_image_pair(imgs[0], imgs[1])
    plot_piv_image_pair(imgs[len(imgs)-2], imgs[len(imgs)-1])
    save_piv_video(image_seq=imgs, output_file="output/PSS_test_case.mp4",fps=32)

test_PSS()
```

### 2.3.Test the events

```python
from generator.flow import FlowManager
from generator.images import config_events, Sequence2events
from utils.visualizer import plot_event_time2d
import time
import numpy as np

def testS2E():
    # flow = FlowManager(mode="data", data_path="./data/backstep_Re1000_00004.npz")
    flow = FlowManager(mode="analytical", name="uniform", uc=4.05,vc=0.00)
    # flow = FlowManager(mode="analytical", name="cellular_flow", vmax=10, p=256)

    flow_func = flow.evaluate
    
    cfg = config_events()
    s2e = Sequence2events(flow_func, cfg)
    evts = s2e.forward()
    
    plot_event_time2d(evts)

    # 调用
    times_new = list()
    for i in range(5):
        t1 = time.time()
        evts = s2e.forward()
        times_new.append(time.time() - t1)
    
    print(f"New Torch (s): {np.mean(times_new):.6f} s ± {np.std(times_new):.6f}")

testS2E()
```

## 3.Test the data loader functions

```python
from generator.flow import FlowManager
from generator.images import config_events
from generator.loader import EventManager
from utils.visualizer import plot_event_time2d, save_voxel_grid_video
from addict import Dict

flow = FlowManager(mode="analytical", name="cellular_flow", vmax=10, p=256)
# flow = FlowManager(mode="analytical", name="uniform", uc=2.5,vc=0.00)

flow_func = flow.evaluate
cfg_evts = config_events() # 支持不同的配置，默认设置见定义
cfg_evts.time_length=15
cfg_evts.d = 2.2


def test_load():
    # 直接读取数据到numpy
    case1 = Dict(source_type="real", path="./data/ebiv_wallflow4_dense_3.h5")
    case2 = Dict(source_type="evk5", path="./data/evk5_rot2.hdf5")
    case3 = Dict(source_type="evk5", path="./data/output11.hdf5")
    case4 = Dict(source_type="npz", path="./data/events.npz")
    # simulater 生成数据 pytorch
    case5 = Dict(source_type="simulator", path=None, cfg=cfg_evts, flow_func=flow_func)

    print("Processing... Most of the time (~5 minutes) will be spent generating the video.")
    for k, case in enumerate([case1, case2, case3, case4, case5]):
    # for k, case in enumerate([case3, case5]):
    # for k, case in enumerate([case5]):
        event_loader = EventManager(**case)
        evts, (u,v) = event_loader.load()
        plot_event_time2d(evts)
        save_voxel_grid_video(evts, # 比较慢 
                              output_file=f"output/loader_event_case{k+1}.mp4", 
                              img_size=None, fps=10)
    

test_load()
```
